```python
import plotly.express as px
import pandas as pd
```

```python
df = pd.read_csv('tourism_data.csv')
df.head()
```

```python
fig = px.line(df, x = 'Month', y = 'Tourist_Arrivals', title='Monthly Tourist Arrivals')
fig.show()
```

#### Based on above line chart
#### 1. Trend is constantly in uptrend from Jan to July.
#### 2. And in downtrend from July to Dec.
#### 3. In July it is the peak season, as the number of arrivals is maximum.

```python
bar_plot = px.bar(df, x = 'Country', y = 'Tourist_Arrivals', title='Tourist Arrivals by Country')
bar_plot.show()
```

#### Inference based on above bar plot
##### 1. Australia has the highest number of arrivals.
##### 2. USA has the lowest number of arrivals.
##### 3. Australia has 150% more arrivals than the USA.

```python
scatter_plot = px.scatter(df, x = 'Average_Travel_Cost', y = 'Hotel_Bookings', title='Average Travel Costs vs Hotel Bookings')
scatter_plot.show()
```

### Inference based on above Scatter Plot
#### 1. With increase in travel cost the hotel bookings are increasing.
#### 2. With increase in hotel bookings, indicates that price increase shows scarcity, increasing hotel bookings.
#### 3. For aveerage Travel Cost 950 (max) it has highest number of bookings (350).

```python
histogram_plot = px.histogram(df, x = 'Hotel_Price', title='Distribution of Hotel Prices')
histogram_plot.show()
```

```python
violin = px.violin(df, x = 'Season', y = 'Average_Travel_Cost' , box= True, title = 'Seasonal Variation in Travel Cost')
violin.show()
```

### Inference
#### 1. Based on above Violin plot, Winter shows most frequent travel costs (wider).
#### 2. Summer has the highest Avg Travel Cost, showing more demand in Summer and the most expensive season.
#### 3. Outliers show premium and budget travelling.
#### 4. Summer is the peak, showing cyclic tourism.

```python
gantt = px.timeline(df, x_start='Start_Date', x_end='End_Date', y = 'Activity', title='Tourims Iternary')
gantt.show()
```

```python
import plotly.figure_factory as ff
import numpy as np
```

```python
# correlation_matrix = df.select_dtypes([include='int']).corr()
correlation_matrix = df.select_dtypes(include='int').corr()
fig = ff.create_annotated_heatmap(
    z = correlation_matrix.values,
    x = correlation_matrix.columns.tolist(),
    y = correlation_matrix.index.tolist(),
    colorscale='Viridis',
    showscale=True,
    annotation_text=[[f"{val:.2f}" for val in row ] for row in correlation_matrix.values],
    font_colors=["white"]
)

fig.update_layout(
    width = 800,
    height = 800,
    margin = dict(t = 50, l = 100), 
    font = dict(size = 10)
)

fig.show()
```
