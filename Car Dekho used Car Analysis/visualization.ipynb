```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('CAR_DETAILS_FROM_CAR_DEKHO.csv')
df.head()
```

```python
sns.set_style("whitegrid")
```

```python
# Fuel type distribution
sns.countplot(x = 'fuel', hue='fuel', data= df, palette='Set2', legend=False)
plt.title('Fuel Type Distribution')
plt.show()
```

```python
# Selling price distribution

sns.histplot(df['selling_price'], kde=True, bins=30, color='skyblue')
plt.title('Selling Price Distribution')
plt.xlabel('Price (INR)')
plt.show()
```

```python
# Price type by Fuel using box plot.

sns.boxplot(x='fuel', y='selling_price', hue='fuel', data=df, palette='Set1')
plt.title('Price distribution by Fuel type.')
plt.show()
```

```python
# Selling Price by Transmission

sns.boxplot(x='transmission', hue='transmission', y='selling_price', data=df, palette='coolwarm')
plt.title('Selling Price by Transmission Type')
plt.show()
```

```python
# Price by ownership history

sns.violinplot(x='owner', y='selling_price',data=df, hue='owner', palette='Set2' )
plt.title('Price by Owenership')
plt.show()
```

#### First owner cars have higher price compared to all other owner type.
#### Second owner and third owner car type loose value rapidly.
#### First owner cars have more market hold.

```python
correlation = df[['year', 'km_driven', 'selling_price']].corr()
sns.heatmap(correlation, annot=True, cmap='YlGnBu')
plt.title('Feature correlation Heatmap')
plt.show()
```
